```python
def get_graycode(n,lt):
    if n==1:
        lt.append("0")
        lt.append("1")
    else:
        get_graycode(n-1,lt)
        q = 2**(n-1)
        for i in range(q):
            lt.append("1"+lt[q-1-i])
            lt[q-1-i] = "0"+lt[q-1-i]
    return lt

lt = []
print(get_graycode(3,lt))
```
